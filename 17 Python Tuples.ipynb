# Python Tuples




##  Introduction to Tuple

A **tuple** is a collection of items in Python that is **ordered** and **immutable** (cannot be changed after creation).

Tuples are written using **parentheses `()`**, with items separated by commas.

**Syntax:**
```python
tuple_name = (item1, item2, item3)
```


````python
# A simple example of a tuple
fruits = ("apple", "banana", "cherry")

print(fruits)
````

## 2. Creating a Tuple

Tuples can be created in different ways depending on how many elements they hold.


````python
# A simple tuple with multiple values
numbers = (10, 20, 30, 40)
print(numbers)
````

---

### Why Do We Use Parentheses `()` in Tuples?

Parentheses are used to clearly show that the values belong to a tuple.  
A tuple can also be created without parentheses; the comma `,` makes it a tuple.  
For a single-value tuple, a comma is compulsory: `(10,)`.  
Tuple functions can be directly used on a tuple without extra parentheses.

````python
# Tuple without parentheses
numbers = 10, 20, 30

print(numbers)
print(type(numbers))
````

````python
# Single-value tuple
number = (10,)
print(number)
print(type(number))
````

````python
# A single-element tuple must have a comma after the value
single_value = (5,)
print(single_value)
````

````python
# An empty tuple with no elements
empty_tuple = ()
print(empty_tuple)
````

---

## 3. Accessing Tuple Elements

Each item in a tuple has a fixed position called an **index**.
Indexing starts from `0` for the first element.
You can access an element by placing its index inside square brackets `[]`.


````python
# Accessing a tuple element by its index
colors = ("red", "green", "blue")

print(colors[1])   # Access the second element
````

---

## 4. Tuple Indexing

**Positive indexing** starts from `0` at the first element and moves forward.

**Negative indexing** starts from `-1` at the last element and moves backward.


````python
numbers = (10, 20, 30, 40)

print(numbers[0])    # Access the first element
print(numbers[-1])   # Access the last element
````

## 5. Tuple Slicing

Slicing lets you get a part of a tuple using the format `tuple[start:stop:step]`.

- `start` is where the slice begins (included)
- `stop` is where the slice ends (excluded)
- `step` is how many items to skip


````python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])   # Elements from index 1 to 3
print(numbers[:3])    # First three elements
print(numbers[::2])   # Every second element
````

## 6. Tuple Operations

Tuples support a few simple operations such as joining, repeating, and checking membership.


````python
a = (1, 2, 3)
b = (4, 5, 6)

# Concatenation: joining two tuples
print(a + b)
````

````python
# Repetition: repeating a tuple's elements
print(a * 2)
````

````python
# Membership: checking if a value exists in a tuple
print(2 in a)
````

## 7. Tuple Functions

Python provides several built-in functions that work with tuples.
A few basic ones are shown below. More functions will be covered in a separate file.


````python
numbers = (10, 20, 30, 40)

print(len(numbers))   # Number of elements in the tuple
print(max(numbers))   # Largest value
print(min(numbers))   # Smallest value
print(sum(numbers))   # Sum of all values
````

## 8. Tuple Methods

Tuples have only two built-in methods: `count()` and `index()`.
These will be explored in more detail in a separate file.


````python
numbers = (10, 20, 20, 30)

print(numbers.count(20))   # Counts how many times 20 appears
print(numbers.index(30))   # Finds the index of value 30
````

## 9. Tuple Mutability

Tuples are **immutable**, which means their elements cannot be changed after creation.
Trying to change a tuple element will cause an error.


````python
numbers = (10, 20, 30)

# Trying to change an element will raise a TypeError
numbers[0] = 100
````
