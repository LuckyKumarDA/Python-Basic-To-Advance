# Python Set





### What is a Set?
- Set is an unordered, mutable collection of unique and heterogeneous elements in Python.
- It is generally created using curly braces {} and does not allow duplicate values.

### Why Sets are used
- To store only **unique** values (no duplicates).
- To perform **mathematical set operations** like union, intersection, difference.
- To quickly check whether a value **exists** in a collection.
- To **compare** two collections of data.

```python
# Basic syntax of a Set
fruits = {"apple", "banana", "cherry"}
print(fruits)
print(type(fruits))
```

### Python Set – Key Points

- A Set is an **unordered collection** of elements.
- A Set is **mutable**, which means it can be modified.
- A Set stores only **unique elements** and does not allow duplicates.
- A Set can store **heterogeneous data types** such as integers, strings, floats, and booleans.
- A Set is generally created using **curly braces `{}`**.
- A Set does **not support indexing and slicing**.
- Set elements must be **immutable/hashable**.
- Sets are useful for **removing duplicates, checking membership, and performing mathematical set operations**.

```python
# Duplicate values are automatically removed
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)   # duplicates removed automatically

# Heterogeneous elements allowed
mixed_set = {1, "hello", 3.14, True}
print(mixed_set)
```

## 3. Creating a Set

### Using curly braces `{ }`
The most common way to create a Set is by placing values inside curly braces, separated by commas.


```python
# Set with integers
a = {10, 20, 30, 40}
print(a)
```

```python
# Set with strings
b = {"red", "green", "blue"}
print(b)
```

```python
# Set with floats
c = {1.1, 2.2, 3.3}
print(c)
```

```python
# Set with mixed data types
d = {1, "two", 3.0, True}
print(d)
```

```python
# Creating a Set from multiple values directly
colors = {"red", "yellow", "green", "red"}   # duplicate "red"
print(colors)
```

### Creating an empty Set

This is a very important and commonly confused topic.

- `{}` does **NOT** create an empty Set — it creates an **empty dictionary**.
- To create a truly empty Set, we must use `set()`.

### Why does `{}` create an empty dictionary instead of an empty Set?
Because curly braces `{ }` were originally used for **dictionaries** in Python. When Python sees empty curly braces, it assumes you mean an empty dictionary — since there is no way to tell (from empty braces alone) whether you meant a Set or a dictionary.


```python
empty_dict = {}
empty_set = set()

print(type(empty_dict))   # <class 'dict'>
print(type(empty_set))    # <class 'set'>
```

```python
# Example with numbers
nums = {5, 1, 5, 3, 1, 2}
print(nums)   # only unique numbers remain

# Example with strings
cities = {"Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"}
print(cities)   # only unique city names remain
```

```python
fruits_list = ["apple", "banana", "cherry"]
fruits_set = {"apple", "banana", "cherry"}

print(fruits_list)   # order is preserved, same every time
print(fruits_set)    # order is not guaranteed to match insertion
```

### Example: Finding unique departments

```python
employee_departments = ["Sales", "HR", "Sales", "Finance", "HR", "IT"]
unique_departments = set(employee_departments)
print(unique_departments)
```

## 23. Limitations of Set

- **No indexing** — individual elements cannot be accessed by position.
- **No slicing** — you cannot extract a sub-part of a Set using index ranges.
- **No duplicate elements** — cannot be used when repeated values need to be preserved.
- **Cannot contain mutable elements** — lists, dictionaries, and other regular Sets cannot be stored inside a Set.
- **Cannot be used when positional order is the main requirement** — if the order of items matters for your task, a Set is not the right choice.
