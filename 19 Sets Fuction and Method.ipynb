# Python Set Functions and Methods



---


## Function: 
- A function is called independently, for example len(my_set).

## Method:
- A method is called using the Set object, for example my_set.add(10).

### 1. `len()`

- Returns the number of elements in a set.
- `len()` simply tells us how many unique elements are present.

**Syntax:**
```python
len(set_name)
```


````python
fruits = {"apple", "banana", "mango", "grapes"}

print(len(fruits))   # counts number of elements
````

### 2. `min()`

- Returns the smallest element in a set.

**Syntax:**
```python
min(set_name)
```


````python
marks = {45, 78, 23, 89, 60}

print(min(marks))   # smallest value
````

### 3. `max()`

- Returns the largest element in a set.

**Syntax:**
```python
max(set_name)
```


````python
marks = {45, 78, 23, 89, 60}

print(max(marks))   # largest value
````

### 4. `sum()`

- Returns the total sum of all numeric elements in a set.

**Syntax:**
```python
sum(set_name)
```


````python
sales = {200, 450, 300, 150}

print(sum(sales))   # total of all values
````

### 5. `sorted()`

- Returns a **new sorted list** made from the set elements. It does **not** sort the set itself (sets have no fixed order).
- `sorted()` returns a **list**, not a set. The original set stays exactly the same.

**Syntax:**
```python
sorted(set_name)
sorted(set_name, reverse=True)
```


````python
scores = {56, 12, 89, 34, 70}

ascending = sorted(scores)
descending = sorted(scores, reverse=True)

print("Ascending :", ascending)
print("Descending:", descending)
print("Original set (unchanged):", scores)
````

###  `set()` — Conversion / Creation Function

- `set()` converts other iterables (list, tuple, string, dictionary keys, etc.) into a set. It automatically removes duplicates.

**Syntax:**
```python
set(iterable)
```


````python
# List -> Set
a  = [1, 2, 2, 3, 4, 4, 5]
b = set(a)
print(b)
````

````python
# Tuple -> Set
my_tuple = (10, 20, 20, 30, 40)
tuple_to_set = set(my_tuple)
print(tuple_to_set)
````

````python
# String -> Set (each character becomes an element)
my_string = "hello"
string_to_set = set(my_string)
print(string_to_set)
````

````python
# Other iterable (e.g., range) -> Set
range_to_set = set(range(5))
print(range_to_set)
````

## Python Set Methods

Set methods are functions that belong **specifically** to the set object and are called using `set_name.method_name()`.

### ➕ Adding Elements

#### 1. `add()`

**What it does:** Adds a **single** element to the set.

**Syntax:**
```python
set_name.add(element)
```


````python
students = {"Amit", "Riya", "John"}
print("Before:", students)

students.add("Sara")   # adding one element
print("After :", students)
````

#### 2. `update()`

**What it does:** Adds **multiple** elements at once (from a list, tuple, another set, etc.).

**Syntax:**
```python
set_name.update(iterable)
```


````python
students = {"Amit", "Riya", "John"}
print("Before:", students)

students.update(["Sara", "Kabir", "Amit"])   # duplicates auto-ignored
print("After :", students)
````

### ➖ Removing Elements

#### 3. `remove()`

**What it does:** Removes a specific element. **Raises an error** (`KeyError`) if the element does not exist.

**Syntax:**
```python
set_name.remove(element)
```


````python
colors = {"red", "green", "blue"}
print("Before:", colors)

colors.remove("green")
print("After :", colors)
````

#### 4. `discard()`

**What it does:** Removes a specific element. Does **NOT** raise an error if the element does not exist — it simply does nothing.

**Syntax:**
```python
set_name.discard(element)
```


````python
colors = {"red", "green", "blue"}
print("Before:", colors)

colors.discard("green")     # exists -> removed
colors.discard("yellow")    # doesn't exist -> no error, nothing happens
print("After :", colors)
````

#### 5. `pop()`

**What it does:** Removes and returns a **random/arbitrary** element from the set (sets are unordered, so you cannot control which one gets removed).

**Syntax:**
```python
set_name.pop()
```


````python
numbers = {10, 20, 30, 40}
print("Before:", numbers)

removed_item = numbers.pop()
print("Removed:", removed_item)
print("After :", numbers)
````

#### 6. `clear()`

**What it does:** Removes **all** elements, leaving an empty set. The set variable still exists.

**Syntax:**
```python
set_name.clear()
```


````python
data = {1, 2, 3, 4}
print("Before:", data)

data.clear()
print("After :", data)
print("Type still set:", type(data))
````

### 🧮 Set Mathematical Operations

````python
dataset_A = {"apple", "banana", "mango", "grapes"}
dataset_B = {"banana", "mango", "orange", "kiwi"}
````

#### 7. `union()` (operator: `|`)

**What it does:** Combines elements from both sets (no duplicates).

**Syntax:**
```python
set1.union(set2)
set1 | set2
```


````python
result_method = dataset_A.union(dataset_B)
result_operator = dataset_A | dataset_B

print(result_method)
print(result_operator)
print(dataset_A)
````

#### 8. `intersection()` (operator: `&`)

**What it does:** Returns elements common to **both** sets.

**Syntax:**
```python
set1.intersection(set2)
set1 & set2
```


````python
common = dataset_A.intersection(dataset_B)
common_op = dataset_A & dataset_B

print(common)
print(common_op)
````

#### 9. `difference()` (operator: `-`)

**What it does:** Returns elements that are in the **first** set but **not** in the second.

**Syntax:**
```python
set1.difference(set2)
set1 - set2
```


````python
only_in_A = dataset_A.difference(dataset_B)

only_in_A_op = dataset_A - dataset_B

print(only_in_A)
print(only_in_A_op)
````

#### 10. `symmetric_difference()` (operator: `^`)

**What it does:** Returns elements that are in **either** set, but **not** in both (i.e., everything except the common elements).

**Syntax:**
```python
set1.symmetric_difference(set2)
set1 ^ set2
```


````python
sym_diff = dataset_A.symmetric_difference(dataset_B)
sym_diff_op = dataset_A ^ dataset_B

print(sym_diff)
print(sym_diff_op)
````

### 🔄 Assignment / Update Methods (These MODIFY the Original Set)

#### 11. `intersection_update()`

**What it does:** Updates the set to keep **only** the common elements (same result as `intersection()`, but changes the original set instead of returning a new one).

**Syntax:**
```python
set1.intersection_update(set2)
```


````python
A = {"apple", "banana", "mango", "grapes"}
B = {"banana", "mango", "orange"}

print(A)
A.intersection_update(B)
print(A)   # A is permanently changed
````

#### 12. `difference_update()`

**What it does:** Updates the set by removing elements that are also present in another set.

**Syntax:**
```python
set1.difference_update(set2)
```


````python
A = {"apple", "banana", "mango", "grapes"}
B = {"banana", "mango"}

print(A)
A.difference_update(B)
print(A)
````

#### 13. `symmetric_difference_update()`

**What it does:** Updates the set with the symmetric difference (keeps elements not common to both).

**Syntax:**
```python
set1.symmetric_difference_update(set2)
```


````python
A = {"apple", "banana", "mango"}
B = {"banana", "orange"}

print(A)
A.symmetric_difference_update(B)
print(A)
````

##   Data Analytics Examples


````python
# Sample datasets used throughout this section
a = {"C101", "C102", "C103", "C104", "C105"}
b = {"C103", "C104", "C105", "C106", "C107"}

cities = ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai", "Chennai"]

categories = ["Electronics", "Grocery", "Electronics", "Clothing", "Grocery"]

sales_data_store1 = {"Laptop", "Mobile", "Tablet", "Headphones"}
sales_data_store2 = {"Mobile", "Tablet", "Smartwatch", "Speaker"}
````

**1. Find unique customers**

````python
all_customers = a.union(b)
print(all_customers)
print(len(all_customers))
````

**2. Find unique cities**

````python
unique_cities = set(cities)
print("Unique cities:", unique_cities)
````

**3. Remove duplicate categories**

````python
unique_categories = set(categories)
print("Unique categories:", unique_categories)
````

**4. Find customers common to two datasets (2025 and 2026)**

````python
returning_customers = a.intersection(b)
print("Customers present in both years:", returning_customers)
````

**5. Find customers present in Dataset A (2025) but not Dataset B (2026) — i.e., lost customers**

````python
lost_customers = a.difference(b)
print("Customers who did NOT return in 2026:", lost_customers)
````

**6. Find new customers (present in 2026 but not in 2025)**

````python
new_customers = b.difference(a)
print("New customers in 2026:", new_customers)
````

**7. Compare two years of customer data (everything that changed)**

````python
changed_customers = a.symmetric_difference(b)
print("Customers that appear in only ONE of the two years:", changed_customers)
````

**8. Find common products between two sales datasets (store 1 and store 2)**

````python
common_products = sales_data_store1.intersection(sales_data_store2)
print("Products sold in BOTH stores:", common_products)
````

**9. Check whether one category list is completely contained in another**

````python
festive_categories = {"Electronics", "Clothing"}
all_categories = {"Electronics", "Grocery", "Clothing", "Toys"}

print("Is festive_categories fully inside all_categories?",
      festive_categories.issubset(all_categories))
````

**10. Find values that exist only in one dataset (unique to store 1, unique to store 2)**

````python
only_store1 = sales_data_store1.difference(sales_data_store2)
only_store2 = sales_data_store2.difference(sales_data_store1)

print("Only in Store 1:", only_store1)
print("Only in Store 2:", only_store2)
````
