# Python Dictionary 



### What is a Dictionary?
A dictionary is a mutable collection of key-value pairs used to store and access data using unique keys. It is written using curly braces {}.

### Key and Value
- **Key** → A unique name used to identify the data (like a label).
- **Value** → The actual data connected to that key.

### Key-Value Pair
A **key** and its **value** together are called a **key-value pair**.

Example: `"name": "Rahul"` → here `"name"` is the key and `"Rahul"` is the value.

### Basic Syntax
```
dictionary_name = {
    key1: value1,
    key2: value2
}
```

Dictionary is written using **curly brackets `{ }`**, and each key is separated from its value using a **colon `:`**.

````python
                                # Simple example of a dictionary
student = {"name": "Lucky Kumar","age": 20,"course": "Data Analyst"}

# Print the whole dictionary
print(student)
````

**Output Explanation:**
The dictionary `student` stores three key-value pairs: name, age, and course. When we print it, Python shows all keys and their values together inside curly brackets.


## Characteristics of Dictionary


1. **Stores data in key-value pairs** – Every value is connected to a unique key.
2. **Keys are unique** – No two keys can be the same in one dictionary.
3. **Values can be duplicated** – Different keys can have the same value.
4. **Dictionary is mutable** – We can change, add, or delete data after creating it.
5. **Dictionary is ordered (in modern Python)** – From Python 3.7 onwards, dictionaries remember the order in which items were added.
6. **Keys must be hashable** – Only unchangeable (immutable) data types like string, number, or tuple can be used as keys. A list cannot be used as a key.
7. **Values can have different data types** – A value can be a string, number, list, tuple, set, or even another dictionary.


````python
# Example showing unique keys and duplicate values
marks = {"maths": 90,"science": 90,"english": 85}           # same value as maths, but that's okay

print(marks)
# Keys (maths, science, english) are unique
# Values (90, 90, 85) can repeat
````

##  Creating a Dictionary

There are many ways to create a dictionary. Let's see simple examples of each.


````python
# 1. Empty dictionary
empty_dict = {}

print(empty_dict)
print(type(empty_dict))
````

````python
# 2. Dictionary with string keys
person = {"name": "Anjali","city": "Delhi"}

print(person)
````

````python
# 3. Dictionary with integer keys
roll_number_names = {101: "Aman",102: "Riya",103: "Karan"}

print(roll_number_names)
````

````python
# 4. Dictionary with different data types as values
mixed_data = {
    "name": "Sonia",       # string value
    "age": 22,              # integer value
    "height": 5.4,           # float value
    "is_student": True       # boolean value
}

print(mixed_data)
````

````python
# 5. Creating a dictionary using dict() function
car = dict(brand="Honda", model="City", year=2022)

print(car)
````

##  Accessing Dictionary Values

We can get the value of a key by writing the key name inside **square brackets `[ ]`**.

### Syntax
```
dictionary_name[key]
```

If we try to access a key that **does not exist**, Python gives an error called **KeyError**.


````python
# Accessing values using keys
student = {"name": "Rahul","age": 20,"course": "B.Com"}

# Access the value of 'name'
print(student["name"])

# Access the value of 'age'
print(student["age"])
````

````python
# What happens when a key does not exist
student = {
    "name": "Rahul",
    "age": 20
}

# Trying to access a key that is not present in the dictionary
print(student["marks"])   # This will give a KeyError
````

**Output Explanation:**
Since the key `"marks"` does not exist in the `student` dictionary, Python raises a **KeyError**. This tells us that the key we are looking for is not present.


##  Adding Data to a Dictionary

We can add a new key-value pair simply by assigning a value to a new key.

### Syntax
```
dictionary_name[new_key] = new_value
```


````python
# Adding a new key-value pair
student = {"name": "Rahul","age": 20}

# Add a new key called 'course'
student["course"] = "B.Com"

print(student)
````

````python
# Adding multiple pieces of information one by one
employee = {"name": "Simran"
}

# Adding more details step by step
employee["department"] = "Sales"
employee["salary"] = 35000

print(employee)
````

##  Updating Data in a Dictionary

We can change the value of an existing key by simply assigning a new value to it.

If the key **already exists**, its value gets updated.
If the key **does not exist**, a new key-value pair gets added.


````python
# Updating an existing value
student = {"name": "Rahul","age": 20}

# Update the age
student["age"] = 21

print(student)
````

````python
# Adding a new key while updating
student = {"name": "Rahul","age": 21}

# 'grade' key does not exist yet, so it will be added
student["grade"] = "A"

print(student)
````

##  Deleting Data from a Dictionary

We can remove data from a dictionary using the **`del`** keyword.

- To delete **one key-value pair**, mention the key name.
- To delete the **complete dictionary**, use `del` on the dictionary name itself.


````python
# Deleting one key-value pair
student = {"name": "Rahul","age": 21,"grade": "A"}

# Delete the 'grade' key
del student["grade"]

print(student)
````

````python
# Deleting the complete dictionary
student = {"name": "Rahul","age": 21}

# Delete the whole dictionary
del student

# Trying to print student now will give an error because it no longer exists
# print(student)
print("Dictionary deleted successfully")
````

##  Nested Dictionary

### What is a Nested Dictionary?
A **nested dictionary** is a dictionary inside another dictionary.

It is useful when we want to store detailed information about something, like a student's personal details along with their marks.

### Simple Example


````python
# Nested dictionary example
student = {"name": "Ravi","age": 20,"marks": {"maths": 85,"science": 90}}

print(student)
````

````python
# Accessing a value from a nested dictionary
student = {"name": "Ravi","marks": {"maths": 85,"science": 90}}

# First access 'marks', then access 'maths' inside it
print(student["marks"]["maths"])
````

````python
# Updating a value inside a nested dictionary
student = {"name": "Ravi","marks": {"maths": 85,"science": 90}}

# Update the maths marks inside the nested dictionary
student["marks"]["maths"] = 95

print(student)
````

````python
# Employee information
employee = {"employee_id": 5001,"name": "Arjun Mehta","department": "Finance","salary": 45000}

print(employee)
````

##  Common Beginner Mistakes

1. **Forgetting commas** between key-value pairs.
   ```
   # Wrong
   student = {"name": "Rahul" "age": 20}
   ```

2. **Incorrect key-value syntax** – forgetting the colon `:` between key and value.
   ```
   # Wrong
   student = {"name" "Rahul"}
   ```

3. **Using duplicate keys** – Python will only keep the last value for a repeated key.
   ```
   student = {"name": "Rahul", "name": "Aman"}
   print(student)   # Output: {'name': 'Aman'}
   ```

4. **Trying to access a key that does not exist** – this gives a `KeyError`.
   ```
   student = {"name": "Rahul"}
   print(student["age"])   # Error: 'age' key not found
   ```

5. **Confusing keys with values** – remember, you always search using the **key**, not the value.

6. **Using an unhashable object (like a list) as a key** – this gives a `TypeError`.
   ```
   # Wrong - list cannot be used as a key
   student = {[1, 2, 3]: "marks"}
   ```


````python
# Example: Duplicate key mistake
student = {
    "name": "Rahul",
    "name": "Aman"   # duplicate key
}

# Python keeps only the last value for a repeated key
print(student)
````
