# Python Dictionary Functions and Methods


##  Dictionary Functions

A function is a reusable block of code that performs a specific task.

### `len()`

- len() returns the number of key-value pairs in a dictionary.
**Syntax:**
```python
len(dictionary)
```

````python
student = {"name": "Lucky Kumar","age": 20,"course": "Data Analyst"
}

# Count the number of key-value pairs
print(len(student))
````

### `type()`
`type()` tells us the data type of a variable. It is useful to check if something is really a dictionary.

**Syntax:**
```python
type(dictionary)
```

````python
student = {"name": "Rahul","age": 20}

# Check the type of the student variable
print(type(student))
````

### `str()`

`str()` converts a dictionary into a string form. This is useful when we want to print or save a dictionary as text.

**Syntax:**
```python
str(dictionary)
```

````python
student = {"name": "Rahul","age": 20}

# Convert the dictionary into a string
student_str = str(student)

print(student_str)
print(type(student_str))
````

### `dict()`

`dict()` is used to create a new dictionary.

**Syntax:**
```python
dict(key1=value1, key2=value2)
```

````python
# Create a new dictionary using dict()
employee = dict(name="Sonia", department="Sales", salary=25000)

print(employee)
````

## 2. Dictionary Methods

A method is a function that belongs to an object and is called using dot (.) notation.

### `get()`

`get()` returns the value of a given key. If the key is not found, it does not give an error; it returns `None` (or a default value if you provide one).

**Syntax:**
```python
dictionary.get(key)
```

````python
student = {"name": "Rahul","age": 20}

# Get the value of the "name" key
print(student.get("name"))

# Try to get a key that does not exist
print(student.get("city"))
````

### `keys()`
`keys()` returns all the keys present in a dictionary.

**Syntax:**
```python
dictionary.keys()
```

````python
student = {"name": "Rahul","age": 20,"course": "B.Com"}

# Get all keys from the dictionary
print(student.keys())
````

### `values()`

`values()` returns all the values present in a dictionary.

**Syntax:**
```python
dictionary.values()
```

````python
# Get all values from the dictionary
print(student.values())
````

### `items()`

`items()` returns all key-value pairs in a dictionary.

**Syntax:**
```python
dictionary.items()
```

````python
# Get all key-value pairs
print(student.items())
````

### `update()`

`update()` adds a new key-value pair to a dictionary, or updates the value if the key already exists.

**Syntax:**
```python
dictionary.update({key: value})
```

````python
# Add a new key-value pair to the dictionary
student.update({"city": "Bijnor"})

print(student)
````

### `pop()`

`pop()` removes a specified key from a dictionary and returns its value.

**Syntax:**
```python
dictionary.pop(key)
```

````python
# Remove the "age" key and return its value
print(student.pop("age"))

print(student)
````

### `popitem()`

`popitem()` removes the last inserted key-value pair from a dictionary and returns it.

**Syntax:**
```python
dictionary.popitem()
```

````python
# Remove the last inserted key-value pair
print(student.popitem())

print(student)
````

### `clear()`

`clear()` removes all key-value pairs from a dictionary. The dictionary becomes empty.

**Syntax:**
```python
dictionary.clear()
```

````python
# Remove all key-value pairs from the dictionary
student.clear()

print(student)
````

### `fromkeys()`

- fromkeys() creates a new dictionary using the given keys and a common value.
**Syntax:**
```python
dict.fromkeys(keys, value)
```

````python
subjects = ["Math", "Science", "English"]

# Create a dictionary with the same starting value (0) for each key
marks = dict.fromkeys(subjects, 0)

print(marks)
````
