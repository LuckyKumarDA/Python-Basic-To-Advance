# Python `range()` and `random` Module


---
##  `range()`

##  Definition

`range()` is a built-in Python function that creates a **sequence of numbers**.

It does not store all the numbers in memory at once. It generates them one by one when needed.

##  Syntax

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

| Term | Meaning | Default |
| ---- | ------- | ------- |
| `start` | The number where counting begins | `0` |
| `stop` | The number where counting ends (**not included**) | Required |
| `step` | The gap between two numbers | `1` |

##  `range(stop)`

Counting starts from `0` and goes up to `stop - 1`.

````python
list(range(5))        # not include stop value
````

````python
list(range(10))     # not include stop value
````

##  `range(start, stop)`

Counting starts from `start` and goes up to `stop - 1`.

````python
# Print numbers from 2 to 5
for i in range(2, 6):
    print(i)
````

##  `range(start, stop, step)`

`step` decides how much to increase (or decrease) each time.

````python
# Print even numbers from 2 to 10
for i in range(2, 11, 2):
    print(i)
````

##  Why Is `stop` Excluded?

The `stop` value is **never included** in the result.

* `range(5)` gives `0, 1, 2, 3, 4` (not 5)
* `range(1, 4)` gives `1, 2, 3` (not 4)



````python
# Convert range to a list to see the values clearly
print(list(range(1, 4)))
````

## Positive Step (Forward Counting)

A positive `step` counts **forward** (numbers get bigger).

````python
# Count forward from 1 to 9 with a step of 2
print(list(range(1, 10, 2)))
````

##  Negative Step (Backward Counting)

A negative `step` counts **backward** (numbers get smaller).

For backward counting, `start` must be **bigger** than `stop`.

````python
# Count backward from 5 to 1
for i in range(5, 0, -1):
    print(i)
````

````python
# Count backward from 10 to 2 with a step of -2
print(list(range(10, 1, -2)))
````

##  `range()` with `list()`

If you print a `range` directly, you only see the range object. Use `list()` to see the numbers.

````python
# Printing range directly does not show the numbers
print(range(5))

# Convert range to a list to see the numbers
print(list(range(5)))
````

##  `range()` with `len()`

`len()` tells you how many numbers are in a range.
It is also used to loop through the positions (indexes) of a list.

````python
# Count how many numbers are in the range
print(len(range(1, 10, 2)))
````

##  Common Mistakes with `range()`

**Mistake 1: Expecting the `stop` value to be included**

````python
# range(5) stops at 4, not 5
print(list(range(5)))

# To include 5, use stop = 6
print(list(range(6)))
````

**Mistake 2: Using a positive step to count backward**

````python
# Wrong: this gives an empty list because the step is positive
print(list(range(5, 0)))

# Correct: use a negative step
print(list(range(5, 0, -1)))
````

**Mistake 3: Using a step of `0`**

`range(1, 5, 0)` gives an error because the step can never be zero.

**Mistake 4: Using decimal numbers**

`range(0, 1, 0.1)` gives an error because `range()` only works with **whole numbers (integers)**.

##  Important Points

* `range()` works only with **integers**.
* `start` is included, `stop` is **not** included.
* Default `start` is `0` and default `step` is `1`.
* `step` can be positive or negative, but never `0`.
* Use `list(range(...))` to see all the numbers.
* `range()` is mostly used with `for` loops.

# Python `random` Module

##  Definition

The `random` module is a built-in Python module that helps us **generate random numbers** and **make random choices**.

## Why Do We Use It?

* To create sample or test data
* To pick random items from a list
* To shuffle data
* To simulate simple things like dice or coin toss


##  Import the Module

You must import the module before using it.

````python
# Import the random module
import random
````

##  `random.random()`

**Purpose:** Generates a random decimal number.

**Syntax:**

```python
random.random()
```

**Return value:** A decimal (float) number from `0.0` up to (but not including) `1.0`.

````python
# Generate a random decimal number between 0 and 1
print(random.random())
````

##  `random.randint()`

**Purpose:** Generates a random **integer** between two numbers.

**Syntax:**

```python
random.randint(start, stop)
```

**Inclusive range:** Both `start` and `stop` can be returned.

````python
# Generate a random integer from 1 to 10 (both 1 and 10 are possible)
print(random.randint(1, 10))
````

##  `random.uniform()`

**Purpose:** Generates a random **decimal** number between two numbers.

**Syntax:**

```python
random.uniform(start, stop)
```

````python
# Generate a random decimal number between 10 and 20
print(random.uniform(10, 20))
````

##  `random.choice()`

**Purpose:** Selects **one** random item from a list.

**Syntax:**

```python
random.choice(sequence)
```

````python
# Select one random fruit
fruits = ["Apple", "Banana", "Mango"]
print(random.choice(fruits))
````

## `random.choices()`

**Purpose:** Selects **multiple** random items from a list.

**Repetition is possible:** The same item can be selected more than once.

**Syntax:**

```python
random.choices(sequence, k=number_of_items)
```

````python
# Select 3 random fruits (repeats are allowed)
fruits = ["Apple", "Banana", "Mango"]
print(random.choices(fruits, k=3))
````

##  `random.sample()`

**Purpose:** Selects **multiple unique** items from a list.

**Repetition is not allowed:** Each item can be selected only once.

**Syntax:**

```python
random.sample(sequence, k)
```

````python
# Select 2 different fruits (no repeats)
fruits = ["Apple", "Banana", "Mango"]
print(random.sample(fruits, 2))
````

##  `random.seed()`

**What is a seed?**
A seed is a starting value for the random number generator.

**Why is it useful?**
If you use the **same seed**, you get the **same random results** every time. This is called *reproducible* results, and it is very useful when you want to share your work or test your code.

**Syntax:**

```python
random.seed(number)
```

````python
# Set a seed to get reproducible results
random.seed(10)

print(random.randint(1, 10))
````

````python
# Using the same seed again gives the same result
random.seed(10)
print(random.randint(1, 10))

random.seed(10)
print(random.randint(1, 10))
````

## Select a Random Product

`random.choice()` picks one item from the list.

````python
# Select a random product
products = ["Laptop", "Phone", "Tablet"]

print(random.choice(products))
````

#  Common Mistakes

## Mistake 1: Forgetting to Import `random`

```python
# Wrong: random is not imported
print(random.randint(1, 10))   # NameError: name 'random' is not defined
```

**Fix:** Always write `import random` at the top of your code.

## Mistake 2: Confusing `randint()` and `randrange()`

`randint(1, 5)` can return `5`, but `randrange(1, 5)` can **never** return `5`.

````python
# randint includes the stop value
print(random.randint(1, 5))    # Can be 1, 2, 3, 4, or 5

# randrange excludes the stop value
print(random.randrange(1, 5))  # Can be 1, 2, 3, or 4
````
