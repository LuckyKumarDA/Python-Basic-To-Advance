## 1. Packing

- **Packing** means putting several values together into a single variable.

- Think of putting many items into one bag. In Python, when you give several values to one variable, Python "packs" them together into one - group. The group can be a **tuple**, a **list** or a **dictionary**.

### Packing values into a tuple
- If we write values separated by commas, Python packs them into a tuple automatically.

```python
# Three values separated by commas are packed into one tuple
numbers = 10, 20, 30

print(numbers)        # Show the packed tuple
print(type(numbers))  # Check the type of the variable
```

###  Packing separate variables into one variable
- Here we have three different variables. We pack them into a single variable called `student`.

```python
name = "Asha"
age = 20
city = "Delhi"

# Pack the three variables into one tuple
student = name, age, city

print(student)  # Show the packed values
```

###  Packing values into a list
- Square brackets `[ ]` can also pack values together. A list can be changed later.

```python
# Pack three colors into one list
colors = ["red", "green", "blue"]

print(colors)       # Show the whole list
print(len(colors))  # Count how many items are packed inside
```

###  Packing values into a dictionary
- A dictionary packs values as `key: value` pairs, so each value has a label.

```python
# Pack related information using key: value pairs
person = {"name": "Ravi", "age": 25, "city": "Mumbai"}

print(person)          # Show the whole dictionary
print(person["name"])  # Get one value using its key
```

## 2. Unpacking

- **Unpacking** means taking values out of a group and storing each value in its own variable.


- Unpacking is the opposite of packing. If packing is putting items into a bag, unpacking is taking the items out one by one. The number of - variables on the left side must be the same as the number of values in the group.

### Unpacking a tuple
- The three values of the tuple are stored in three separate variables.

```python
# A tuple with three values
point = (10, 20, 30)

# Unpack the tuple: each value goes to one variable, in order
x, y, z = point

print(x)  # First value
print(y)  # Second value
print(z)  # Third value
```

###  Unpacking a list
- Unpacking works with lists in the same way.

```python
# A list with three fruits
fruits = ["apple", "banana", "mango"]

# Unpack the list into three variables
first, second, third = fruits

print("First fruit:", first)
print("Second fruit:", second)
print("Third fruit:", third)
```

###  Swapping two variables
- Unpacking lets us swap the values of two variables in a single line.

```python
a = 5
b = 8
print("Before swap:", a, b)

# The right side is packed first, then unpacked into a and b
a, b = b, a

print("After swap:", a, b)
```

## 3. Extended Unpacking

- **Extended unpacking** is used when we want to unpack multiple values, but we do not want to create a separate variable for every value.

- We use the `*` operator to collect the remaining values into a list.

### Collecting middle values



- `first` gets the first value.
- `*middle` collects all remaining middle values into a list.
- `last` gets the last value.

```python
# A tuple with five values
numbers = (10, 20, 30, 40, 50)

# Extended unpacking
first, *middle, last = numbers

print("First:", first)
print("Middle:", middle)
print("Last:", last)
```

###  Collecting remaining values

The `*` operator can collect multiple remaining values into a list.

```python
# A list of fruits
fruits = ["apple", "banana", "mango", "orange", "grapes"]

# Extended unpacking
first, *remaining = fruits

print("First fruit:", first)
print("Remaining fruits:", remaining)
```
