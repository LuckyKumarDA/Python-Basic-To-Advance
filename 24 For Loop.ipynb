# Python `for` Loop 


### What is a loop?
- A **loop** is a way to repeat a task many times without writing the same code again and again.

### What is a `for` loop?
- A **`for` loop** repeats a block of code once for every item in a sequence. A sequence is a collection of things, such as a list of - numbers, a list of names, or the letters of a word.

### Why do we use a `for` loop?
- Imagine you need to print 100 numbers. Without a loop, you would write 100 `print` lines. With a `for` loop, you write just a few lines and Python does the repeating for you.

### Very basic Python example

````python
# A list of days of the week (a sequence of items)
days = ["Monday", "Tuesday", "Wednesday"]

# The loop takes one day at a time and stores it in the variable 'day'
for day in days:
    print("Today is", day)  # This line runs once for each day
````

---

## 2. `for` Loop Syntax

The basic syntax of a `for` loop is:

```python
for variable in sequence:
    # code to repeat
```

### Simple Meaning

* `for` → starts the loop
* `variable` → stores one value at a time
* `in` → takes values from the sequence
* `sequence` → list, string, range, etc.
* `:` → starts the loop body
* indentation → tells Python which code belongs to the loop

````python
### Simple Example

for i in range(1, 6):
    print(i)
````

Here, `i` takes each value from `1` to `5`, one by one, and `print(i)` prints the value.

### Remember


The `for` loop takes values **one by one** from the sequence and runs the code for each value.

````python
# range(1, 6) gives the numbers 1, 2, 3, 4, 5 (6 is NOT included)
for i in range(1, 6):
    print(i)  # Print numbers from 1 to 5
````

## 3. `for` Loop with `range()`

The `range()` function creates a sequence of numbers. It is very useful with a `for` loop.

**Important:** In `range()`, the **stop value is NOT included**. For example, `range(1, 5)` gives 1, 2, 3, 4. The number 5 is not included.

### `range(stop)`
Starts at 0 by default and goes up to (but not including) `stop`.

````python
# Starts at 0, goes up to 4 (5 is NOT included)
for number in range(5):
    print(number)  # Prints 0, 1, 2, 3, 4
````

### `range(start, stop)`
Starts at `start` and goes up to (but not including) `stop`.

````python
# Starts at 2, goes up to 6 (6 is NOT included)
for number in range(2, 6):
    print(number)  # Prints 2, 3, 4, 5
````

### `range(start, stop, step)`
The `step` tells Python how much to add each time. The default step is 1.

````python
# Starts at 0, goes up to 10 (10 is NOT included), and adds 2 each time
for number in range(0, 10, 2):
    print(number)  # Prints 0, 2, 4, 6, 8
````

## 4. `for` Loop with Numbers


### Example 1: Print numbers from 1 to 5

````python
# Loop through numbers 1 to 5 (stop value 6 is not included)
for number in range(1, 6):
    print(number)  # Print the current number
````

### Example 2: Print numbers from 1 to 10

````python
# Loop through numbers 1 to 10 (stop value 11 is not included)
for number in range(1, 11):
    print(number)  # Print the current number
````

### Example 3: Print numbers from 10 to 1

To count backwards, we use a **negative step**. Here, `-1` means "subtract 1 each time". The stop value is `0` because it is not included.

````python
# Start at 10, go down to 1 (stop value 0 is not included), step is -1
for number in range(10, 0, -1):
    print(number)  # Print 10, 9, 8, ..., 1
````

### Example 4: Print even numbers from 1 to 10

Even numbers are numbers divisible by 2 (2, 4, 6, 8, 10). We start at 2 and add 2 each time.

````python
# Start at 2, stop before 11, and add 2 each time
for number in range(2, 11, 2):
    print(number)  # Print 2, 4, 6, 8, 10
````

### Example 5: Print odd numbers from 1 to 10

Odd numbers are numbers not divisible by 2 (1, 3, 5, 7, 9). We start at 1 and add 2 each time.

````python
# Start at 1, stop before 11, and add 2 each time
for number in range(1, 11, 2):
    print(number)  # Print 1, 3, 5, 7, 9
````

## 5. `for` Loop with a List

A **list** is a collection of items stored in square brackets `[ ]`, separated by commas.

When a `for` loop works with a list, it takes **one item at a time** from the list. On the first loop, the variable holds the first item. On the next loop, it holds the second item, and so on, until all items are used.

````python
# A list of fruits
fruits = ["Apple", "Banana", "Mango"]

# 'fruit' takes one item from the list each time
for fruit in fruits:
    print(fruit)  # Print each fruit
````

````python
# Example 1: A list of student names
students = ["Riya", "Aman", "Priya"]

for student in students:
    print("Welcome,", student)  # Greet each student
````

````python
# Example 2: A list of prices
prices = [100, 250, 75]

for price in prices:
    print("Price:", price)  # Print each price
````

````python
# Example 3: A list of cities
cities = ["Delhi", "Mumbai", "Chennai", "Kolkata"]

for city in cities:
    print("City:", city)  # Print each city name
````

## 6. `for` Loop with a String

A **string** is a piece of text, such as `"Lucky"`. A string is also a sequence, because it is made of characters (letters, numbers, spaces, symbols).

A `for` loop can go through a string and give you **one character at a time**.

````python
# A string with 5 characters
name = "Lucky"

# 'char' takes one character from the string each time
for char in name:
    print(char)  # Print each character
````

## 7. Simple Calculations Using a `for` Loop

We can use a `for` loop to do calculations. The idea is to keep a **running total** and add to it during each loop.

### Example 1: Sum of numbers from 1 to 5

````python
total = 0  # Start the total at 0

# Loop through numbers 1 to 5
for number in range(1, 6):
    total = total + number  # Add the current number to the total

print("Sum of numbers from 1 to 5 is:", total)  # Expected: 15
````

### Example 2: Sum of numbers from 1 to 10

````python
total = 0  # Start the total at 0

# Loop through numbers 1 to 10
for number in range(1, 11):
    total = total + number  # Add the current number to the total

print("Sum of numbers from 1 to 10 is:", total)  # Expected: 55
````

### Example 3: Multiplication table of 5

````python
# Loop from 1 to 10 to create the table
for i in range(1, 11):
    result = 5 * i  # Multiply 5 by the current number
    print(f"5 x {i} = {result}")  # Print one line of the table
````

The `f"..."` text is called an **f-string**. Anything inside `{ }` is replaced by its value. So `{i}` becomes the current number and `{result}` becomes the answer.

## 9. Common Beginner Mistakes

### Mistake 1: Forgetting the colon `:`

This code has an error because the colon is missing at the end of the `for` line:

```python
for number in range(1, 4)
    print(number)
```

**Corrected version:**

````python
# The colon ':' at the end of the for line fixes the error
for number in range(1, 4):
    print(number)
````

### Mistake 2: Wrong indentation

This code has an error because `print` is not indented under the loop:

```python
for number in range(1, 4):
print(number)
```

**Corrected version:**

````python
# The print line is indented, so it belongs to the loop
for number in range(1, 4):
    print(number)
````

### Mistake 3: Confusing `range(1, 5)` with `range(1, 6)`

Many beginners want the numbers 1 to 5, but they write `range(1, 5)`. That gives only 1, 2, 3, 4, because the stop value is not included.

```python
# Wrong: prints 1, 2, 3, 4 (5 is missing)
for number in range(1, 5):
    print(number)
```

**Corrected version:** use `range(1, 6)` to get 1, 2, 3, 4, 5.

````python
# Correct: prints 1, 2, 3, 4, 5
for number in range(1, 6):
    print(number)
````
