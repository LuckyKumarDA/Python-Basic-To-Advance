# Python Comprehensions 





## 1. List Comprehension

- A list comprehension is a short and easy way to create a list by writing a `for` loop in a single line, inside square brackets `[]`.

###  Basic List Comprehension

```python
# Create a list of numbers from 1 to 5 using a normal for loop
numbers = []
for num in range(1, 6):
    numbers.append(num)
print(numbers)
```

```python
# Now do the same thing using list comprehension
numbers = [num for num in range(1, 6)]
print(numbers)
```

```python
# Example: create a list of squares of numbers from 1 to 5
squares = [num * num for num in range(1, 6)]
print(squares)
```

###  List Comprehension with a Condition

- We can add an `if` condition at the end of a list comprehension to include only the items that satisfy the condition.

```python
# Create a list of only even numbers from 1 to 10
even_numbers = [num for num in range(1, 11) if num % 2 == 0]
print(even_numbers)
```

```python
# Create a list of words that have more than 3 letters
fruits = ["fig", "apple", "kiwi", "banana", "pear"]
long_fruits = [fruit for fruit in fruits if len(fruit) > 3]
print(long_fruits)
```

### List Comprehension with if-else

- We can use `if-else` inside a list comprehension to choose between two values for every item. Here the `if-else` comes right after the expression, before the `for` part.

```python
# Label each number as "even" or "odd"
numbers = [1, 2, 3, 4, 5]
labels = ["even" if num % 2 == 0 else "odd" for num in numbers]
print(labels)
```

```python
# Replace negative numbers with 0, keep positive numbers as they are
values = [4, -2, 7, -9, 0, 3]
positive_values = [num if num >= 0 else 0 for num in values]
print(positive_values)
```

## 2. Set Comprehension

- A set comprehension is just like a list comprehension, but it creates a **set** instead of a list, using curly braces `{}`. A set never has duplicate values.

### 2.1 Simple Example

```python
# Create a set of squares from a list that has duplicate numbers
numbers = [1, 2, 2, 3, 3, 4]
square_set = {num * num for num in numbers}
print(square_set)
```

```python
# Create a set of the first letter of each word
words = ["cat", "car", "dog", "duck"]
first_letters = {word[0] for word in words}
print(first_letters)
```

### 2.2 Example with Condition

- Just like list comprehension, we can add an `if` condition in a set comprehension to include only certain items.

```python
# Create a set of even numbers only (duplicates are removed automatically)
numbers = [1, 2, 2, 3, 4, 4, 5, 6]
even_set = {num for num in numbers if num % 2 == 0}
print(even_set)
```

```python
# Create a set of vowels found in a word
word = "beginner"
vowels = {letter for letter in word if letter in "aeiou"}
print(vowels)
```

## 3. Dictionary Comprehension

**Definition:** A dictionary comprehension is a short way to create a dictionary using `{key: value for item in sequence}` syntax.

### 3.1 Simple Example

```python
# Create a dictionary that maps each number to its square
numbers = [1, 2, 3, 4, 5]
square_dict = {num: num * num for num in numbers}
print(square_dict)
```

```python
# Create a dictionary that maps each fruit name to the length of its name
fruits = ["apple", "fig", "kiwi"]
fruit_lengths = {fruit: len(fruit) for fruit in fruits}
print(fruit_lengths)
```

### 3.2 Example with Condition

**Definition:** We can add an `if` condition in a dictionary comprehension to include only the key-value pairs that satisfy the condition.

```python
# Create a dictionary of squares, but only for even numbers
numbers = [1, 2, 3, 4, 5, 6]
even_square_dict = {num: num * num for num in numbers if num % 2 == 0}
print(even_square_dict)
```

```python
# Create a dictionary of student marks, keeping only students who passed (marks >= 40)
marks = {"Amit": 55, "Riya": 30, "John": 70, "Sara": 20}
passed_students = {name: score for name, score in marks.items() if score >= 40}
print(passed_students)
```
