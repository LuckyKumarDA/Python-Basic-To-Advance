# Python Jump Statements

Jump statements are used inside loops to change the normal flow of the loop.

In Python, there are three jump statements:

1. `break`
2. `continue`
3. `pass`

## 1. `break`


- `break` is used to stop the loop immediately, even if the loop has not finished all its rounds.

- We use `break` when we want to exit a loop as soon as a certain condition becomes true. It is helpful when we don't need to check the remaining values.

**Basic Syntax:**
```python
for item in sequence:
    if condition:
        break
```

````python
# Example 1: Stop the loop when i becomes 3
for i in range(1, 6):
    if i == 3:
        break  # Stop the loop when i becomes 3
    print(i)
````

````python
# Example 2: Search for a number in a list and stop once found
numbers = [4, 7, 2, 9, 5]

for num in numbers:
    if num == 9:
        print("Found 9!")
        break  # Stop searching once the number is found
    print("Checking:", num)
````

## 2. `continue`

- `continue` is used to skip the current round of the loop and move to the next round.

We use `continue` when we want to skip certain values but still continue running the rest of the loop.

**Basic Syntax:**
```python
for item in sequence:
    if condition:
        continue
```

````python
# Example 1: Skip the number 3
for i in range(1, 6):
    if i == 3:
        continue  # Skip printing when i is 3
    print(i)
````

````python
# Example 2: Print only odd numbers
for i in range(1, 8):
    if i % 2 == 0:
        continue  # Skip even numbers
    print(i)
````

````python
# Example 3: Skip a specific name in a list
names = ["Amit", "Riya", "Skip", "Sam"]

for name in names:
    if name == "Skip":
        continue  # Skip this name
    print(name)
````

## 3. `pass`

- `pass` is used when a statement is required but we do not want to do anything at that moment. It does nothing and simply moves on.

- We use `pass` as a placeholder, usually when we are writing the structure of code first and will add the actual logic later. Without `pass`, Python will show an error for an empty block.

**Basic Syntax:**
```python
for item in sequence:
    if condition:
        pass
```

````python
# Example 1: Using pass as a placeholder inside a loop
for i in range(1, 6):
    if i == 3:
        pass  # Do nothing for now, just move on
    print(i)
````

````python
# Example 3: Using pass inside an if-else block
x = 10

if x > 5:
    pass  # Placeholder, decide the logic later
else:
    print("x is small")

print("Code continues normally")
````

**Simple way to remember:**
- `break` = **Stop** the loop.
- `continue` = **Skip** this round, but keep looping.
- `pass` = **Do nothing**, just move on (placeholder).
