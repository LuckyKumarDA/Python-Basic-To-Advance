# Python Functions 

---

## 1. What is a Function?
- A function is a block of organized, reusable code that performs a specific task.

- Instead of writing the same code again and again, we write it once inside a function and use it whenever we need it.



## 2. Why do we use Functions?

- Functions help us organize code, avoid repeating ourselves, and make programs easier to read and fix.

**Main reasons to use functions:**
- **Reusability** – Write the code once, use it many times.
- **Readability** – Code becomes clean and easy to understand.
- **Easy Debugging** – Easier to find and fix mistakes in small, separate blocks.
- **Saves Time** – No need to rewrite the same logic again and again.

## 3. Creating a Function using `def`

- In Python, we create (define) a function using the `def` keyword, followed by the function name and parentheses `()`.

**Basic Syntax:**
```python
def function_name():
    # code to execute
```

````python
# Creating a simple function named 'greet'
def greet():
    print("Hello, welcome to Python!")   # this line runs only when the function is called

# Note: Just defining a function does not run it.
# We have not called it yet, so nothing is printed here.
````

## 4. Calling a Function

- Calling a function means using its name followed by parentheses `()` to actually run the code written inside it.

````python
# Calling the greet function defined above
greet()
````

## 5. Function without Parameters

- A function without parameters does not take any input. It simply performs a fixed task every time it is called.

````python
def say_hello():
    print("Hello! Have a nice day.")           # always prints the same message
    
say_hello()
````

## 6. Function with One Parameter

- A parameter is a value that we pass into a function so it can use that value to do its task. A function with one parameter accepts exactly one input.

````python
def greet_person(name):     # 'name' is the parameter
    print("Hello,", name)

greet_person("Riya")            # "Riya" is passed as input
````

## 7. Function with Multiple Parameters

- A function can accept more than one parameter. Multiple parameters are separated using commas.

````python
def add_numbers(num1, num2):    # two parameters: num1 and num2
    result = num1 + num2
    print("Sum is:", result)

add_numbers(5, 3)
````

## 8. Parameters vs Arguments


- **Parameter** – The variable name written inside the function definition.
- **Argument** – The actual value we pass when calling the function.

````python
def multiply(a, b):     # a and b are PARAMETERS
    print(a * b)

multiply(4, 5)            # 4 and 5 are ARGUMENTS passed to the function
````

## 9. Positional Arguments

- Positional arguments are values passed to a function in the same order as the parameters are defined. The order matters here.

````python
def student_info(name, age):
    print(name, "is", age, "years old")

student_info("Aman", 20)   # "Aman" goes to 'name', 20 goes to 'age' (order matters), 
````

## 10. Keyword Arguments

- Keyword arguments are passed using the parameter name. Since we mention the name, the order does not matter.

````python
def student_info(name, age):
    print(name, "is", age, "years old")

student_info(age=20, name="Aman")   # order changed, but still works correctly
````

## 11. Default Parameters

- A default parameter has a preset (ready-made) value. If we do not pass a value for it, Python uses the default value.

````python
def greet_user(name="Guest"):   # "Guest" is the default value
    print("Hello,", name)

greet_user("Priya")   # overrides the default value
greet_user()          # no value passed, so default value "Guest" is used
````

## 12. `return` Statement

- The `return` statement sends a value back from the function to the place where it was called, so that value can be stored and used later.

````python
def square(num):
    return num * num    # sends the result back to the caller

result = square(4)      # 'result' stores the value returned by the function
print(result)
````

## 13. Difference between `print()` and `return`

- `print()` only **displays** a value on the screen. It does not give the value back to the program.
- `return` **gives the value back**, so it can be stored in a variable and used further.

````python
def add_print(a, b):
    print(a + b)        # only shows the result on screen, gives nothing back

def add_return(a, b):
    return a + b         # returns the result so it can be stored

add_print(2, 3)                 # output is shown directly
value = add_return(2, 3)        # 'value' now stores the returned result
print("Stored value:", value)
````

## 14. Returning Multiple Values

- A Python function can return more than one value at the same time. The values are separated by commas.

````python
def calculate(a, b):
    add = a + b
    sub = a - b
    return add, sub        # returning two values together

sum_result, sub_result = calculate(10, 4)   # unpacking both values
print("Sum:", sum_result)
print("Difference:", sub_result)
````

## 15. `*args`

- `*args` allows a function to accept **any number** of positional arguments. Python collects all the values into a group called a tuple.

````python
def add_all(*numbers):        # can take any number of arguments
    total = 0
    for n in numbers:         # loop through each value one by one
        total = total + n
    print("Total:", total)

add_all(1, 2, 3)
add_all(5, 10, 15, 20)
````

## 16. Function with `if-else`

- We can use `if-else` conditions inside a function to make decisions based on the input value.

````python
def check_even_odd(num):
    if num % 2 == 0:          # check if number is exactly divisible by 2
        print(num, "is Even")
    else:
        print(num, "is Odd")

check_even_odd(7)
check_even_odd(10)
````

## 17. Function with a Loop

- We can use loops like `for` or `while` inside a function to repeat a task multiple times.

````python
def print_table(num):
    for i in range(1, 6):        # loop runs 5 times (1 to 5)
        print(num, "x", i, "=", num * i)

print_table(3)
````

### Student Marks and Grade

- We create a function that takes a student's marks and returns their grade using `if-else` conditions.

````python
def calculate_grade(marks):
    if marks >= 90:
        grade = "A"
    elif marks >= 75:
        grade = "B"
    elif marks >= 50:
        grade = "C"
    else:
        grade = "Fail"
    return grade          # send the grade back

student_marks = 82
result_grade = calculate_grade(student_marks)
print("Marks:", student_marks)
print("Grade:", result_grade)
````

### Simple Calculator

- We create a function that works like a basic calculator. It takes two numbers and an operation name, then performs the correct calculation.

````python
def calculator(num1, num2, operation):
    if operation == "add":
        return num1 + num2
    elif operation == "subtract":
        return num1 - num2
    elif operation == "multiply":
        return num1 * num2
    elif operation == "divide":
        return num1 / num2
    else:
        return "Invalid operation"

print(calculator(10, 5, "add"))         # 10 + 5
print(calculator(10, 5, "subtract"))    # 10 - 5
print(calculator(10, 5, "multiply"))    # 10 * 5
print(calculator(10, 5, "divide"))      # 10 / 5
````
