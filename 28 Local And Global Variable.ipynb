# Local Variable and Global Variable in Python



## 1. Local Variable

- A **local variable** is a variable that is created **inside a function**. It can only be used **inside that function**. We cannot use it outside the function.

```python
def greet():
    # This is a local variable. It is created inside the function.
    message = "Hello, I am a local variable!"
    
    # We can use the local variable inside the same function.
    print(message)

# Calling the function to run the code inside it
greet()
```

**What happens if we try to use it outside the function?**

Since `message` is a local variable, Python will give an error if we try to print it outside the function.

```python
# Trying to use the local variable outside the function
# This will give an error because 'message' only exists inside greet()
print(message)
```

## 2. Global Variable

- A **global variable** is a variable that is created **outside any function**. It can be used **anywhere in the program**, both inside and outside functions.

```python
# This is a global variable. It is created outside any function.
city = "Delhi"

def show_city():
    # We can use the global variable inside a function too.
    print("I live in", city)

# Calling the function
show_city()

# We can also use the global variable outside the function
print("City is:", city)
```

```python
# Global Variable

name = "Lucky"

def show_name():
    print(name)  # Using global variable

show_name()
```

```python
# Local Variable

def student():
    age = 20  # Local variable
    print(age)

student()
```

```python
# Global and Local Variable Together

city = "Dehradun"  # Global variable

def student():
    course = "Python"  # Local variable
    
    print(city)
    print(course)

student()
```

```python
# Same Variable Name

name = "Lucky"  # Global variable

def student():
    name = "Rahul"  # Local variable
    print(name)

student()

print(name)
```

## Differene

- A **local variable** lives **inside** a function and can only be used there.
- A **global variable** lives **outside** functions and can be used **everywhere** in the program.

```python
s = 10
def xyz(a,b):
    global s,p
    s = 100
    p = 1000
    return a+b+s+p

```
