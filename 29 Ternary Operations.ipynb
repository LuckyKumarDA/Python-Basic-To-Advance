# Ternary Operator in Python

## What is a Ternary Operator?

A **ternary operator** is a short way to write a simple `if-else` condition in a single line.

It is used when you want to choose between two values based on a condition.

## Basic Syntax

```python
value_if_true if condition else value_if_false
```

- If the **condition** is `True`, it returns **value_if_true**.
- If the **condition** is `False`, it returns **value_if_false**.

## Check Even or Odd

````python
num = 10

# Check whether the number is even or odd
result = "Even" if num % 2 == 0 else "Odd"

print(result)
````

## Check Pass or Fail

````python
marks = 65

# Check whether the student passed or failed
result = "Pass" if marks >= 40 else "Fail"

print(result)
````

## Example 3: Find Greater Number

````python
a = 25
b = 18

# Find the greater number
greater = a if a > b else b

print("Greater number:", greater)
````

````python
# Store a number
num = -5

# Check whether the number is positive or negative
result = "Positive" if num >= 0 else "Negative"

print(result)
````

````python
# Store age
age = 20

# Check eligibility
result = "Eligible" if age >= 18 else "Not Eligible"

print(result)
````

## Ternary Operator vs if-else

**Using normal if-else:**

````python
num = 7

# Normal if-else to check even or odd
if num % 2 == 0:
    result = "Even"
else:
    result = "Odd"

print(result)
````

**Same logic using ternary operator:**

````python
num = 7

# Same check using ternary operator
result = "Even" if num % 2 == 0 else "Odd"

print(result)
````

# Ternary Operator with Multiple Conditions


When we have multiple conditions, we can use a **Nested Ternary Operator**.

### Syntax

```python
value1 if condition1 else value2 if condition2 else value3

````python
salary = 45000

# Check salary category using multiple conditions
category = "High" if salary >= 60000 else "Medium" if salary >= 30000 else "Low"

# Print the final category
print(category)
````

````python
units = 350

# Check the electricity usage level
usage = "Very High" if units > 500 else "High" if units > 400 else "Medium" if units > 300 else "Low"

# Print the usage level
print(usage)
````

Both examples give the same result, but the ternary operator writes it in just one line.

## Important Points

- The ternary operator is used for **simple conditions**.
- It makes short `if-else` statements more **compact**.
- For **complex conditions**, a normal `if-else` is easier to read and understand.
