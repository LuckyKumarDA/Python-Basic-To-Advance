# Python Lambda Functions



##  What is a Lambda Function?

> A lambda function is a small anonymous function that is written in one line using the `lambda` keyword.

**What does "anonymous" mean?**
A lambda function is called *anonymous* because it does not need a name (unlike normal functions created with `def`). You can create it and use it right away, without giving it a formal name.

**What does `lambda` mean?**
`lambda` is just a keyword in Python. It tells Python: "I am creating a small function here."

##  Basic Syntax


```python
lambda arguments: expression
```



                                  
- `lambda`      Keyword used to create a lambda function 
- `arguments`   Input values given to the function       
- `:`           Separates arguments from the expression  
- `expression`  Operation that returns the result        

A lambda function can take any number of arguments, but it can only have **one expression**. That expression is automatically returned as the result — there is no need to write `return`.

## Simple Example



````python
# Create a lambda function that adds 10
add_10 = lambda x: x + 10

# Call the lambda function
print(add_10(5))
````

**Let's understand every line:**

* `add_10 = lambda x: x + 10` — This line creates a lambda function and stores it inside the variable `add_10`. So now, `add_10` is a function.
* `lambda x` — This means the function takes one input, and we are calling that input `x`.
* `x + 10` — This is the expression. It takes the input `x` and adds 10 to it. This result is automatically returned.
* `add_10(5)` — This calls the function and passes `5` as the value of `x`.
* Inside the function, Python calculates `5 + 10`, which gives `15`.
* So `print(add_10(5))` prints `15` on the screen.

##  Lambda with Two Arguments

A lambda function can take more than one argument. Let's create one that adds two numbers.

````python
# Create a lambda function to add two numbers
add = lambda a, b: a + b

# Call the function
print(add(10, 20))
````

**Let's understand every line:**

* `add = lambda a, b: a + b` — This creates a lambda function with two inputs, `a` and `b`, and stores it in the variable `add`.
* `lambda a, b` — This means the function needs two values to work.
* `a + b` — This expression adds the two inputs together and returns the result.
* `add(10, 20)` — This calls the function with `a = 10` and `b = 20`.
* Python calculates `10 + 20`, which gives `30`.
* So `print(add(10, 20))` prints `30`.

## Lambda for Multiplication

Now let's create a lambda function that multiplies two numbers.

````python
# Create a lambda function to multiply two numbers
multiply = lambda a, b: a * b

# Call the function
print(multiply(5, 4))
````

**Let's understand every line:**

* `multiply = lambda a, b: a * b` — This creates a lambda function with two inputs, `a` and `b`, and stores it in the variable `multiply`.
* `a * b` — This expression multiplies the two inputs and returns the result.
* `multiply(5, 4)` — This calls the function with `a = 5` and `b = 4`.
* Python calculates `5 * 4`, which gives `20`.
* So `print(multiply(5, 4))` prints `20`.

##  Lambda vs Normal Function

Let's compare a normal function and a lambda function that do the exact same job: finding the square of a number.

### Normal Function

````python
# Create a normal function
def square(x):
    return x * x

# Call the function
print(square(5))
````

### Lambda Function

````python
# Create the same function using lambda
square = lambda x: x * x

# Call the function
print(square(5))
````

## Lambda with `if-else`

A lambda function can also make a small decision using `if-else`. Let's check whether a number is even or odd.

````python
# Check whether a number is even or odd
check = lambda x: "Even" if x % 2 == 0 else "Odd"

# Call the lambda function
print(check(10))
````

````python
# Check whether marks are passing or failing
check_marks = lambda marks: "Pass" if marks >= 40 else "Fail"

# Call the lambda function
print(check_marks(75))
````

**Let's understand every part:**

* `%` — This is called the *modulus operator*. It gives the remainder after dividing one number by another. For example, `10 % 2` gives `0`, because 10 divides evenly by 2 with no remainder.
* `x % 2 == 0` — This checks if the remainder is `0`. If it is, the number is even.
* `if` — This is the condition we are checking. If the condition is true, Python picks the value before `if` (in this case `"Even"`).
* `else` — If the condition is false, Python picks the value after `else` (in this case `"Odd"`).
* **How the result is selected:** Python first checks `x % 2 == 0`. Since `10 % 2` is `0`, the condition is true, so the lambda function returns `"Even"`.
* `check(10)` calls the function with `x = 10`, and since 10 is even, the output is `Even`.
