# Python Error Handling


## What is Error Handling?

**Error handling is a way to handle errors in a program without stopping the entire program unexpectedly.**


##  Why Do We Need Error Handling?

Let's look at a simple example. This program asks the user to enter a number and then prints it.

```python
number = int(input("Enter a number: "))  # Ask the user for input and convert it into an integer

print(number)                             # Print the number entered by the user
```

**What happens if the user enters `abc` instead of a number?**

Python will raise a `ValueError`, because the text `"abc"` cannot be converted into an integer. When this happens, the program stops immediately with an error message.

This is exactly the kind of problem error handling helps us solve — we can catch the error and respond to it, instead of letting the whole program crash.

## Basic `try` and `except`

- `try` → contains the code that may cause an error.
- `except` → runs only if an error happens inside the `try` block, and handles that error.

Example:

```python
try:
    number = int(input("Enter a number: "))  # Convert user input into an integer
    print(number)                            # Print the converted number
except ValueError:
    print("Please enter a valid number.")    # Run when the input is not a valid integer
```

##  Handling Division by Zero

Dividing a number by zero is not allowed in Python. It raises a `ZeroDivisionError`. Let's see how to handle it.

```python
try:
    number = 10        # A normal number
    divisor = 0         # A divisor that is zero

    result = number / divisor  # This line will cause an error
    print(result)               # This line will not run because of the error above
except ZeroDivisionError:
    print("Cannot divide by zero.")  # Runs when division by zero happens
```

##  Handling Different Types of Errors

Sometimes more than one type of error can happen in the same `try` block. We can add more than one `except` block to handle each error separately.

```python
try:
    number = int(input("Enter a number: "))  # Convert input into an integer
    result = 100 / number                     # Divide 100 by the number
    print(result)

except ValueError:
    print("Please enter a valid number.")     # Runs if the input is not a number

except ZeroDivisionError:
    print("Number cannot be zero.")           # Runs if the number is zero
```

##  The `else` Block

**The `else` block runs only when no error occurs in the `try` block.**

Example:

```python
try:
    number = int(input("Enter a number: "))  # Convert input into integer
except ValueError:
    print("Invalid input.")                  # Handle invalid input
else:
    print("Valid number:", number)           # Runs when no error occurs
```

##  The `finally` Block

**The `finally` block always runs, whether an error occurs or not.**

Example:

```python
try:
    number = int(input("Enter a number: "))  # Convert input into integer
    print(number)                            # Print the number
except ValueError:
    print("Invalid input.")                  # Handle the error
finally:
    print("Program finished.")              # Always runs, error or not
```

## `try` + `except` + `else` + `finally` Together

Now let's combine everything we have learned into one complete example.

```python
try:
    number = int(input("Enter a number: "))  # Convert input into integer
    result = 100 / number                     # Divide 100 by the number

except ValueError:
    print("Please enter a valid number.")     # Handle invalid input

except ZeroDivisionError:
    print("Number cannot be zero.")           # Handle division by zero

else:
    print("Result:", result)                  # Runs when no error occurs

finally:
    print("Calculation completed.")           # Always runs
```

##  Common Python Errors

- `ValueError`           Wrong value type or format
- `TypeError`            Wrong data type operation
- `ZeroDivisionError`    Division by zero
- `KeyError`             Dictionary key does not exist
- `NameError`            Variable name is not defined

##  Simple `TypeError` Example

A `TypeError` happens when we try to perform an operation on the wrong data type, such as adding text and a number together.

```python
try:
    result = "10" + 5                        # Try to add a string and an integer
    print(result)                            # Print the result
except TypeError:
    print("Cannot add string and integer.")  # Handle TypeError
```

##  Simple `IndexError` Example

An `IndexError` happens when we try to access a position in a list that does not exist.

```python
try:
    numbers = [10, 20, 30]              # Create a list with 3 items (index 0, 1, 2)
    print(numbers[5])                   # Try to access index 5, which does not exist
except IndexError:
    print("Index does not exist.")      # Handle IndexError
```

## Simple `KeyError` Example

A `KeyError` happens when we try to access a dictionary key that does not exist.

```python
try:
    student = {"name": "Lucky", "age": 19}  # Create a dictionary with two keys
    print(student["city"])                  # Try to access a key that does not exist
except KeyError:
    print("Key does not exist.")            # Handle KeyError
```

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    result = a / b
    print("Result:", result)

except Exception as e:
    print("An error occurred:", e)
```

```python
try:
    numbers = [10, 20, 30]
    index = int(input("Enter index: "))
    print("Value:", numbers[index])

except Exception as e:
    print("An error occurred:", e)
```
